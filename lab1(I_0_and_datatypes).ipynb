##Output with single variable

```python
a=5
print("a=",a)
```

##Output with double variable

```python
a,b= 5,6
print(a,b)
print("a",a,"b",b)
print("a "+str(a),"b "+str(b))
print("a = {} b = {}".format(a,b))
print(f"a={a} b={b}")
```
