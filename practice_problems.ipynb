```python
n = int(input("Enter number: "))
even_sum = 0
odd_sum = 0

while n > 0:
    digit = n % 10
    if digit % 2 == 0:
        even_sum = even_sum + digit
    else:
        odd_sum = odd_sum + digit
    n = n // 10

print("Sum of even digits =", even_sum)
print("Sum of odd digits =", odd_sum)
```

```python
n = int(input())
arr_str = input().split()

arr = []
for i in range(n):
    arr.append(int(arr_str[i]))

for i in range(n):
    for j in range(0, n - i - 1):
        if arr[j] > arr[j + 1]:
            temp = arr[j]
            arr[j] = arr[j + 1]
            arr[j + 1] = temp

for i in range(n):
    print(arr[i], end=" ")
print()
```

```python
def swap_positions(lst, pos1, pos2):
    temp = lst[pos1]
    lst[pos1] = lst[pos2]
    lst[pos2] = temp
    return lst

lst = input("Enter list items: ").split()
p1 = int(input("Enter first position: "))
p2 = int(input("Enter second position: "))

print(swap_positions(lst, p1, p2))
```

```python
def get_sum_and_average(lst):
    total = 0
    count = 0

    for item in lst:
        total = total + int(item)
        count = count + 1

    average = total / count
    print("Sum =", total)
    print("Average =", average)

lst = input("Enter numbers separated by space: ").split()
get_sum_and_average(lst)
```

```python
def check_common_member(list1, list2):
    for item1 in list1:
        for item2 in list2:
            if item1 == item2:
                return True
    return False

lst1 = input("Enter first list items: ").split()
lst2 = input("Enter second list items: ").split()
print(check_common_member(lst1, lst2))
```

```python
lst_str = input("Enter numbers separated by space: ").split()
lst = []

for item in lst_str:
    lst.append(int(item))

# Sort the list using a simple bubble sort
for i in range(len(lst)):
    for j in range(0, len(lst) - i - 1):
        if lst[j] > lst[j + 1]:
            temp = lst[j]
            lst[j] = lst[j + 1]
            lst[j + 1] = temp

print("Second smallest =", lst[1])
print("Second largest =", lst[len(lst) - 2])
```
