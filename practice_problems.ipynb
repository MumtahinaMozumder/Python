```python
def swap_positions(lst, pos1, pos2):
    temp = lst[pos1]
    lst[pos1] = lst[pos2]
    lst[pos2] = temp
    return lst

lst = input("Enter list items: ").split()
p1 = int(input("Enter first position: "))
p2 = int(input("Enter second position: "))

print(swap_positions(lst, p1, p2))
```

```python
def get_sum_and_average(lst):
    total = 0
    count = 0

    for item in lst:
        total = total + int(item)
        count = count + 1

    average = total / count
    print("Sum =", total)
    print("Average =", average)

lst = input("Enter numbers separated by space: ").split()
get_sum_and_average(lst)
```

```python
lst_str = input("Enter numbers separated by space: ").split()
lst = []

for item in lst_str:
    lst.append(int(item))

# Sort the list using a simple bubble sort
for i in range(len(lst)):
    for j in range(0, len(lst) - i - 1):
        if lst[j] > lst[j + 1]:
            temp = lst[j]
            lst[j] = lst[j + 1]
            lst[j + 1] = temp

print("Second smallest =", lst[1])
print("Second largest =", lst[len(lst) - 2])
```

```python
import time

def log_time(func):
    def wrapper():
        start_time = time.time()
        func()
        end_time = time.time()
        print("Time taken by function:", end_time - start_time, "seconds")
    return wrapper
@log_time
def slow_function():
    for i in range(1000000):
        pass
    print("Function is done!")

slow_function()
```

```python
def fib_generator(terms):
    a = 0
    b = 1
    for i in range(terms):
        yield a
        c = a + b
        a = b
        b = c
print("First 6 Fibonacci numbers:")
for num in fib_generator(6):
    print(num, end=" ")
print()
```

```python
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id

    def display(self):
        print(f"Student Name: {self.name}")
        print(f"Student ID: {self.student_id}")

s1 = Student("Alice", "CS101")
s2 = Student("Bob", "CS102")

s1.display()
s2.display()
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)
rect = Rectangle(5, 3)
print("Length:", rect.length, "Width:", rect.width)
print("Area =", rect.area())
print("Perimeter =", rect.perimeter())
```

```python
class BookInventory:
    def __init__(self):
        self.books = []

    def add_book(self, book_name):
        self.books.append(book_name)
        print(f"'{book_name}' has been added.")

    def remove_book(self, book_name):
        if book_name in self.books:
            self.books.remove(book_name)
            print(f"'{book_name}' has been removed.")
        else:
            print("Book not found in inventory.")

    def display_inventory(self):

        for book in self.books:
            print("-", book)
my_library = BookInventory()

my_library.add_book("Python Crash Course")
my_library.add_book("Clean Code")
my_library.display_inventory()

my_library.remove_book("Clean Code")
my_library.display_inventory()
```
