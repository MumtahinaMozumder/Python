#3. Largest of Two Numbers
Problem: Print the larger of two numbers.

7 12
Output:
12


```python
a=int(input())
b=int(input())
if a>b:
  print(f"{a}")
elif b>a:
  print(f"{b}")
else:
  print(f'{a} and {b} equal')
```

Problem: Calculate simple interest.
Formula: SI = (P × R × T) / 100
Input:
1000 5 2
Output:
100


```python
P,R,T=map(int,input().split())
SI=(P*R*T)/100
print(f"SI ={SI}")
```
